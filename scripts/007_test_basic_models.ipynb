```python
from models import ClimateModel, BaseModel, LinearRegression, BoostingModel, create_nn, MLPModel, LSTMModel, TCNModel, UNetModel, load_nn
from train import make_train_test, evaluate_print

FIRST_YEAR = 1991
SPLIT_YEAR = 2016
LAST_YEAR = 2019

def make_ds(element, args):
    return make_train_test(element, FIRST_YEAR, SPLIT_YEAR, LAST_YEAR, args)
```

```python
variables = ['hice', 'sst', 'aice', 'cld', 'h500', 'olr', 'prec', 'ps', 'rq2', 't2', 't2max', 't2min', 't850', 'u850', 'uv10',
 'v850', 'cos_lat', 'sin_lon', 'cos_lon']
dsTrain, dsTest = make_ds('aice', args={"lead_times": [3], "periods": [3], "normed": True,
                                        "cache_level": 2, "batch_size": 1e8, "variables": variables.copy()})
models = [ClimateModel(), BaseModel(), LinearRegression(variables=variables), BoostingModel(variables=variables),
         create_nn('aice', MLPModel,
            {'input_size': len(dsTrain.variables), 'hidden_layers': [4, 2] },
            test=dsTest, lr=0.04, epochs=200, verbose=False, loss_type='acc')]

print('Доля морского льда, март')
for model in models:
    model.fit(dsTrain)
    evaluate_print(model, dsTrain, dsTest)

models = [ClimateModel(), BaseModel(), LinearRegression(variables=variables), BoostingModel(variables=variables),
         create_nn('aice', MLPModel,
            {'input_size': len(dsTrain.variables), 'hidden_layers': [4, 2] },
            test=dsTest, lr=0.04, epochs=200, verbose=False, loss_type='acc')]

print('Доля морского льда, август')
dsTrain, dsTest = make_ds('aice', args={"lead_times": [3], "periods": [8], "normed": True,
                                        "cache_level": 2, "batch_size": 1e8, "variables": variables.copy()})
for model in models:
    model.fit(dsTrain)
    evaluate_print(model, dsTrain, dsTest)
```
