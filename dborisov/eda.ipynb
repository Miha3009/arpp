```python
import netCDF4 as nc 
import matplotlib.pyplot as plt
import numpy as np
import itertools
import xarray as xr
import glob
from pathlib import Path
import torch
import patcher
import pandas as pd
```

```python
DATA_DIR = Path('/Volumes/portable/arpp/data')
patcher_context = patcher.Context(str(DATA_DIR), 16)
```

```python
ALL_VARS = [f.stem for f in list(DATA_DIR.glob('*.bin'))]
INM_VARS = [v for v in ALL_VARS if v.split("_")[0] == 'inm']
ERA_VARS = [v for v in ALL_VARS if v.split("_")[0] != 'inm']
ERA_VARS.remove('config')
print(f"ERA_VARS:\n {' / '.join(ERA_VARS)}")
print(f"INM_VARS:\n {' / '.join(INM_VARS)}")
```

#### **INMCM сравнение netcdf/bin**

```python
conv_dct = {'PREC':'tp', 
            'PS': 'mslp', 
            'SFR': 'snow_cover', 
            'SS': 'swe', 
            'T2': 't2m'}

def draw_inm(date, month, var): 
    tag   = int(month[4:6])
    ens_memb = 0

    fl = DATA_DIR / var / f'north{month}.nc'

    ds = nc.Dataset(fl)
    arr_nc = ds[var][:]
    n_ens, n_days, n_lat, n_lon = arr_nc.shape

    dates = list(pd.date_range(f'{month}01', periods=n_days, freq='d').strftime('%Y%m%d'))
    #print(f'dates from {dates[0]} to {dates[-1]}')
    date_idx = dates.index(date)

    arr = np.roll(arr_nc[ens_memb,date_idx], 180)

    var_patcher = ('inm_' + var.lower()) if var not in list(conv_dct.keys()) else ('inm_' + conv_dct[var])
    request = patcher.Request(var_patcher, #element
                            0, #x0
                            0, #y0 
                            date, #t0
                            n_lon, #xSize 
                            n_lat, #ySize
                            1, #tSize 
                            1, #xyStep
                            1, #tStep 
                            tag) #tag
    anom = patcher.load(patcher_context, [request])[0]
    clim = patcher.load_climate(patcher_context, [request])[0]
    anom = anom.squeeze(1) #delete date axis
    #tensor = anom+clim.unsqueeze(0)
    tensor = (anom+clim)[ens_memb] if var_patcher not in ['inm_snow_cover'] else anom[ens_memb]
    print(anom.shape, clim.shape, tensor.shape)

    fig, axes = plt.subplots(1,2)
    fig.set_size_inches(15,5)
    plot1= axes[0].pcolormesh(np.array(tensor))
    plot2 = axes[1].pcolormesh(arr)
    fig.colorbar(plot1, ax=axes[0])
    fig.colorbar(plot2, ax=axes[1])

    fig.tight_layout()

    axes[0].set_title(f'{var} {date} patcher')
    axes[1].set_title(f'{var} {date} from netcdf')

    plt.show()

date = '19910318'
month = '199102'

for var in ['H500', 'HLT', 'OLR', 'PREC', 'PS', 'SFR', 'SS', 'T2', 'TS', 'U850', 'V850', 'WW']: 
    draw_inm(date, month, var)
```

```python
def get_tensor(var, date, mode = 'abs', tStep=1): 
    n_lon = 1440 if var[:4] != 'inm_' else 360
    n_lat = 361 if var[:4] != 'inm_' else 91
    tag = int(date[4:6]) if var[:4] == 'inm_' else 0
    print(tag)
    request = patcher.Request(var, #element
                                    0, #x0
                                    0, #y0 
                                    date, #t0
                                    n_lon, #xSize 
                                    n_lat, #ySize
                                    1, #tSize 
                                    1, #xyStep
                                    tStep, #tStep 
                                    tag,#month
                                    )
    
    anom = patcher.load(patcher_context, [request])[0]
    print(anom.shape)
    clim = patcher.load_climate(patcher_context, [request])[0]
    tensor = (anom+clim).squeeze() if var not in ['pt', 'snow_cover', 'inm_snow_cover'] else anom.squeeze()
    tensor = tensor[0] if var[:4] == 'inm_' else tensor
    tensor = tensor.numpy()
    print(tensor.shape)

    #tensor = np.where(tensor > 300.0, np.nan, tensor) if var == 'sd' else tensor 
    if mode == 'abs': 
        return tensor
    elif mode == 'anom':
        return anom[0].squeeze().numpy()
    elif mode == 'clim':
        return clim.squeeze().numpy()

def draw(tensor): 
    plt.pcolormesh(tensor)
    plt.colorbar()
    plt.show()

def describe_tensor(tensor): 
    s = pd.Series(tensor.flatten()).dropna()
    fig, (ax1, ax2) = plt.subplots(1,2)
    ax1.hist(s, bins=100)
    ax1.axvline(s.median(), color='r', linewidth = 0.5)
    ax1.axvline(s.mean(), color='g', linewidth = 0.5)
    ax2.boxplot(s, meanline = True, showfliers=False)
    plt.show()
    desc = s.describe(percentiles=[0.01, 0.25, 0.5, 0.75, 0.99, 0.999])
    desc['zeros'] = ((s == 0.0).sum() / desc['count']) * 100
    return desc

glaicer = get_tensor('glaicer', '19930106')
lsm = get_tensor('lsm', '20000606')
alt = get_tensor('z', '20000606')

_var = 'sden'
date = '20180301'
mode = 'anom' 

var = get_tensor(_var, 
                date,
                mode=mode,
                tStep=7
                )
#var_clim = get_tensor(_var, date, mode='clim')
if _var == 'sden': 
#    result = np.where((glaicer != 0.0) | (lsm == 0.0) | (alt >= 2999.0), np.nan, var)
    result = np.where((var <= 0.001) & (var>= -0.001), np.nan, var)
#    if mode == 'abs': 
#        result = np.where((result <= 101.0) | (result >= 299.9), np.nan, result)
#    else: 
#        result = np.where((result <= 0.001) & (result >= -0.001), np.nan, result)
    #result = np.where(result==100.0, 0, result)
elif _var == 'sd': 
    result = np.where((glaicer != 0.0) | (lsm == 0.0) | (alt >= 2999.0), np.nan, var)
    result = np.where((result <= 0.001) & (result>= -0.001), np.nan, result)
    #result = np.where(result==0.0, np.nan, result)
    #result = np.log1p(result)
    #result = np.where((np.abs(var) <= 0.01), np.nan, var)
elif _var == 'inm_swe': 
    result = np.where((var <= 0.01) & (var >= -0.01), np.nan, var)
elif _var[-2:] == 'tp' and mode=='abs':
    result = np.where(var < 0.0, 0.0, var)
    result = np.log1p(result)
#elif _var == 'inm_swe': 
#    result = np.where((var <= 0.001) | (var >= 500) , np.nan, var)
elif _var == 'snow_cover': 
    result = np.where((glaicer != 0.0) | (lsm == 0.0) | (var == 0.0), np.nan, var)
elif _var == 'inm_snow_cover': 
    result = np.where((var == 0.0), np.nan, var)
elif _var == 'inm_ww': 
    #result = np.where((var <= 0.01) & (var >= -0.01), np.nan, var)
    result = np.where((var == 0.0), np.nan, var)
    #result = np.log1p(result)
elif _var in ['sst', 'inm_hlt'] : 
    result = np.where((var <= 0.5) & (var>= -0.5), np.nan, var)
    #result = np.where((var <= 1) & (var >= -1), np.nan, var)
    #result = np.log1p(var)
else: 
    result = var

#result = np.where((var >= 400) | (var <= 1.0), np.nan, var)
#result = np.where((var != 100.0) & (var != 300.0), var, np.nan)
#result = var[220:350, 400:650]

draw(result)
describe_tensor(result)
```

```python
from scripts.utils import ContiguousDataset
from matplotlib import pyplot as plt

train_dates = pd.date_range('19910201', '20201001', freq='MS').strftime('%Y%m%d')

train_ds = ContiguousDataset(
    vars = {'era_vars': ['snow_cover', 'sd', 'sst', 't2m', 'tp', 'sden', 'pt'],
            'inm_vars': ['inm_swe', 'inm_ts', 'inm_mslp', 'inm_tp', 'inm_u850', 'inm_v850', 'inm_snow_cover', 'inm_olr', 'inm_h500', 'inm_hlt', 'inm_t2m'],  
            'surf_vars': ['z', 'sdor', 'lsm', 'glaicer']}, 
    dates = train_dates, 
    scales={'era_scales': 
    [  
        {'id': '6weeks', 'tStep': 7, 'tSize': 6},
        {'id': '6months', 'tStep': 30, 'tSize': 6}
        ]
    , 'inm_scales':
    [
        {'id': '17weeks', 'tSize': 17, 'tStep': 7}, # maximum lead time = 119-122 days, = 17 weeks
    ]})

idx = list(train_dates).index('19930901')
instnc = train_ds[idx]

for key, tensor in instnc.items():
    print(f'{key}: {tensor.shape}')
    if key.startswith('inm_lead_time'): 
        continue
    to_draw = tensor[0,0] if key.startswith('inm') else tensor[-1]
    fig, (ax1, ax2) = plt.subplots(1, 2)
    fig.set_size_inches(15,5)
    mesh = ax1.pcolormesh(to_draw)
    fig.colorbar(mesh, ax=ax1)
    ax1.set_title(key)
    ax2.hist(tensor.flatten(), bins=100)
    ax2.set_title(str(tensor.shape))
```
