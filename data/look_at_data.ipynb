```python
import netCDF4 as nc
import matplotlib.pyplot as plt
```

```python
ds = nc.Dataset('train/swe/anom/199105.nc')
for var in ds.variables:
    print(var, '--', ds[var][:].shape)

#rq - отн. влаж 2м 
#ws - влагозапас почвы 10 м
#ww - 100 м
```

```python
vars_2d = [var for var in ds.variables if ds[var][:].ndim > 1]

for i, var in enumerate(vars_2d, start=1):
    fig,ax = plt.subplots(figsize=(10, 2.5))
    var_to_d = ds[var][0] if ds[var][:].ndim == 3 else ds[var][:]
    im = ax.imshow(var_to_d, origin='lower')
    plt.colorbar(im, ax=ax, shrink = 0.6)
    ax.set_title(f'{var}, {var_to_d.shape}')

plt.tight_layout()
plt.show()
```
